```python
import numpy as np
import matplotlib.pyplot as plt

img_arr = np.zeros((200,200,3),dtype=np.uint8)
img_arr[:, :] = np.full((200, 200, 3), 255, dtype=np.uint8)

img_arr[10:60, 20:90] = [255,0,0];
img_arr[70:120 , 20:90] = [0,255,0];
img_arr[130:180 , 20:90] = [0,0,255];

plt.imshow(img_arr)
```
