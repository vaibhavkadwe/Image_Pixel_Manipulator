```python

import numpy as np
import matplotlib.pyplot as plt

# Project "Image Pixel Manipulator (Synthetic Images)"

'''Intitalizing Image array'''
img = np.zeros((200, 200, 3), dtype=np.uint8)

# Filling the bg with white pixels
img[:, :] = np.full((200, 200, 3), 255, dtype=np.uint8)

img[10:60, 20:90, 0] = 255       #red
img[70:120, 20:90, 1] = 255      #green
img[130:180, 20:90, 2] = 255     #blue

'''Mixing Different Colours'''
img[10:60, 110:180, 0] = 255
img[10:60, 110:180, 1] = 255

img[70:120, 110:180, 0] = 255
img[70:120, 110:180, 2] = 255

#Drawing the lines
img[130:180, 110:180, :] = 0
img[130:180:5, 110:180, :] = 255

'''Display the original image'''
plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.title("Original Synthetic Image")
plt.axis("off")
plt.show()


# 2. Crop the image
crop = img[10:120, 20:90, :]

plt.figure(figsize=(4, 4))
plt.imshow(crop)
plt.title("Cropped Image")
plt.axis("off")
plt.show()


# 3. Flip the image horizontally
flipped = img[:, ::-1, :]

plt.figure(figsize=(5, 5))
plt.imshow(flipped)
plt.title("Horizontally Flipped Image")
plt.axis("off")
plt.show()


# 4. Extract R, G and B channels
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.show()


# 5. Brighten the image
float_img = img.astype(np.float32)

bright = float_img + 50
bright = np.clip(bright, 0, 255).astype(np.uint8)

plt.figure(figsize=(5, 5))
plt.imshow(bright)
plt.title("Brightened Image")
plt.axis("off")
plt.show()


# Darken the image
dark = np.clip(float_img - 50, 0, 255).astype(np.uint8)

plt.figure(figsize=(5, 5))
plt.imshow(dark)
plt.title("Darkened Image")
plt.axis("off")
plt.show()


# 6. Find bright pixels using a Boolean mask
mask = np.any(img > 200, axis=2)

print("Number of bright pixels:", np.sum(mask))

# Show only the bright pixels
bright_pixels = np.zeros_like(img)
bright_pixels[mask] = img[mask]

plt.figure(figsize=(5, 5))
plt.imshow(bright_pixels)
plt.title("Bright Pixels Only")
plt.axis("off")
plt.show()


'''Memory usage Comparison'''
img_uint8 = img
img_float64 = img.astype(np.float64)

print("Image shape:", img.shape)
print("uint8 data type:", img_uint8.dtype)
print("float64 data type:", img_float64.dtype)

print("uint8 memory:", img_uint8.nbytes, "bytes")
print("float64 memory:", img_float64.nbytes, "bytes")



random_img = np.random.randint(
    0, 256, (200, 200, 3), dtype=np.uint8
)

ones_img = np.ones((200, 200, 3), dtype=np.uint8) * 255

constant_img = np.full(
    (200, 200, 3), 128, dtype=np.uint8
)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(random_img)
plt.title("Random Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(ones_img)
plt.title("White Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(constant_img)
plt.title("Gray Image")
plt.axis("off")

plt.show()
```
